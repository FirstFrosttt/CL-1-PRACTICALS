```python
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
# Load the air quality dataset
df = pd.read_csv('City_Air_Quality.csv')

print("Dataset imported successfully!")
print("Dataset Shape:", df.shape)
```

```python
print("First 5 rows:")
display(df.head())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python
print("Available Columns:")
print(df.columns.tolist())

# Select relevant variables
selected_columns = ['Date', 'City', 'PM2.5', 'PM10', 'CO', 'AQI']

print("\nRelevant Variables:")
for col in selected_columns:
    if col in df.columns:
        print(col, "— Available")
    else:
        print(col, "— Not Available")
```

```python
df['Date'] = pd.to_datetime(df['Date'])

# Calculate average AQI for each date across cities
daily_aqi = df.groupby('Date')['AQI'].mean()

# Plot overall AQI trend
plt.figure(figsize=(7, 4))
plt.plot(daily_aqi.index, daily_aqi.values, color='blue', label='Average AQI')

plt.title('Overall AQI Trend Over Time')
plt.xlabel('Date')
plt.ylabel('Average AQI')
plt.legend()
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
pollutants = ['PM2.5', 'PM10', 'CO']

for pollutant in pollutants:
    daily_pollutant = df.groupby('Date')[pollutant].mean()

    plt.figure(figsize=(7, 4))
    plt.plot(daily_pollutant.index, daily_pollutant.values,
             label=pollutant)

    plt.title(f'{pollutant} Levels Over Time')
    plt.xlabel('Date')
    plt.ylabel(f'{pollutant} Level')
    plt.legend()
    plt.xticks(rotation=45)
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
aqi_comparison = daily_aqi.head(10)
plt.figure(figsize=(7, 4))
plt.bar(aqi_comparison.index.strftime('%Y-%m-%d'),
        aqi_comparison.values,
        color='skyblue',
        label='Average AQI')

plt.title('Average AQI Comparison Across Dates')
plt.xlabel('Date')
plt.ylabel('Average AQI')
plt.legend()
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
aqi_data = df.dropna(subset=['AQI', 'AQI_Bucket'])

# Create box plot for each AQI category
plt.figure(figsize=(7,4))

categories = aqi_data['AQI_Bucket'].unique()
box_data = [
    aqi_data[aqi_data['AQI_Bucket'] == category]['AQI']
    for category in categories
]

plt.boxplot(box_data, tick_labels=categories)

plt.title('Distribution of AQI Across Categories')
plt.xlabel('AQI Category')
plt.ylabel('AQI Value')
plt.xticks(rotation=30)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
pollutants = ['PM2.5', 'PM10', 'CO']
for pollutant in pollutants:
    plot_data = df.dropna(subset=['AQI', pollutant])

    plt.figure(figsize=(6, 4))
    plt.scatter(plot_data[pollutant], plot_data['AQI'],
                alpha=0.5, color='green')

    plt.title(f'AQI vs. {pollutant}')
    plt.xlabel(f'{pollutant} Level')
    plt.ylabel('AQI')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.plot(daily_aqi.index, daily_aqi.values,
         color='darkblue', linewidth=2,
         label='Average AQI')

plt.title('City Air Quality Index (AQI) Trends',
          fontsize=16, fontweight='bold')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Average AQI', fontsize=12)

plt.legend(loc='best')
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
