Title :Reinforcement Learning  
Solve the Taxi problem using reinforcement learning where the agent acts as a taxi driver to pick up a passenger at one location and then drop the passenger off at their
destination.

```python
import gymnasium as gym
import numpy as np
```

```python
#Create Taxi environment
env = gym.make("Taxi-v4")
q_table = np.zeros((env.observation_space.n, env.action_space.n))
print("States:", env.observation_space.n)
print("Actions:", env.action_space.n)
```

```python
#Train the agent using Q-Learning
for episode in range(10000):
    state, info = env.reset()
    done = False

    while not done:
        action = env.action_space.sample() if np.random.rand() < 0.1 else np.argmax(q_table[state])

        next_state, reward, terminated, truncated, info = env.step(action)
        done = terminated or truncated

        q_table[state, action] += 0.1 * (
            reward + 0.99 * np.max(q_table[next_state]) - q_table[state, action]
        )

        state = next_state

print("Training completed!")
```

```python
#Test the trained agent
state, info = env.reset()
done = False
total_reward = 0

while not done:
    action = np.argmax(q_table[state])
    state, reward, terminated, truncated, info = env.step(action)
    done = terminated or truncated
    total_reward += reward

print("Total Reward:", total_reward)
env.close()
```

```python
print("Q-Table Shape:", q_table.shape)
print("Taxi problem solved using Reinforcement Learning.")
```
